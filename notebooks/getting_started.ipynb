# UnSynth — getting started

Detect AI-text / watermark *signals* on an already-published article, then run a multi-pass clean. This notebook uses the local heuristic stack (no cloud, no GPU).

```python
from pathlib import Path
from unsynth import UnSynthPipeline, load_settings
from unsynth.types import PipelineMode
from unsynth.eval import render_markdown_report

text = Path('../examples/sample_article.md').read_text()
pipe = UnSynthPipeline(load_settings())
before = pipe.detect(text)
print(before.score, before.label)
for m in before.details['members']:
    print(f"  {m['name']:14} {m['score']:.3f} {m['label']}")
```

```python
result = pipe.run(text, mode=PipelineMode.CLEAN)
print(render_markdown_report(result))
print('---')
print(result.output[:800])
```
